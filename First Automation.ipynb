# Systems and Process Automation with Python
### Challenge:
Every day, our system updates the sales from the previous day.
As an analyst, your daily task is to send an email to the board of directors as soon as you start working, reporting the revenue and the quantity of products sold the day before.

Board of directors' email: management@example.com
Location where the system provides the previous day's sales: https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga

To solve this, we will use pyautogui, a library for automating mouse and keyboard commands.



```python
import pyautogui
import time
pyautogui.PAUSE = 1

# Task Walkthrough

# Step 1: Log in to the company's system (drive link)
# Open Chrome
pyautogui.press ("win")
pyautogui.write ("chrome")
pyautogui.press ("enter")
pyautogui.click(x=684,y=509)
time.sleep (1)

# Type the system link
Link = "https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga"
pyautogui.write(Link)

# Press enter and wait for it to load the system

pyautogui.press("enter")
time.sleep(1)



```

```python
# Step 2: Navigate the system to find the database.
pyautogui.click(x=318, y=321, clicks=2)
time.sleep(1)

# Step 3: Import the database (download the file)
pyautogui.click(x=321, y=331) # Select the file
time.sleep(0.5)
pyautogui.click(x=553, y=243) # Open the options (three dots)
time.sleep(0.5)
pyautogui.click(x=586, y=317) # Download the file
time.sleep(5)
```

### Let's now read the downloaded file to retrieve the indicators:

- Revenue
- Quantity of Products

```python
import pandas

# Step 4: Calculate the indicators (revenue and quantity of products sold)
# Open database
caminho = r"data/Vendas - Dez.xlsx"
tabela = pandas.read_excel(caminho)

# View database information
display(tabela)
# Sum the revenue of all products = sum the Final Value column
faturamento = tabela["Valor Final"].sum()
# Sum the quantity of product = sum the Quantity column
qtde_produtos = tabela["Quantidade"].sum()

print (faturamento)
print (qtde_produtos)
```

### Let's now send an email using Gmail.

```python
import pyperclip

# Step 5: Send the information by email
# Open a new tab
pyautogui.hotkey("ctrl", "t")

# Enter email
pyautogui.write("https://mail.google.com/")
pyautogui.press("enter")
time.sleep(3)

# Click the write button
pyautogui.click(x=63, y=216)
time.sleep(0.5)

# Who should the email be sent to?
pyautogui.write("management@example.com")
pyautogui.press("tab")
pyautogui.press("tab")

# What is the subject of the email?
pyperclip.copy("Relatório de Vendas")
pyautogui.hotkey("ctrl","v")
pyautogui.press("tab")

# Input email body
texto = f"""
Dear all,

Please find today's sales report attached.

Revenue = R${faturamento:,.2f}
Quantity of products sold = {qtde_produtos:,}

Please let me know if you have any questions.

Best regards,
T. Duarte
"""
pyperclip.copy(texto)
pyautogui.hotkey("ctrl","v")
# Enviar o email
pyautogui.click(x=1005, y=1021)
```
